```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully")
```

```python
data = {
    'Date': pd.date_range('2026-11-01', periods=30),
    'Steps': np.random.randint(3000, 12000, 30),
    'Calories': np.random.randint(1500, 2800, 30),
    'Sleep_Hours': np.random.uniform(5, 9, 30),
    'Exercise_Minutes': np.random.randint(10, 90, 30)
}

df = pd.DataFrame(data)

print(df.head())
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())
```

```python
print("First 5 Records:")
print(df.head())

print("\nLast 5 Records:")
print(df.tail())

print("\nRandom 5 Records:")
print(df.sample(5))
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())
```

```python
df['Steps'] = df['Steps'].fillna(df['Steps'].mean())

df['Calories'] = df['Calories'].fillna(df['Calories'].mean())

df['Sleep_Hours'] = df['Sleep_Hours'].fillna(df['Sleep_Hours'].mean())

df['Exercise_Minutes'] = df['Exercise_Minutes'].fillna(df['Exercise_Minutes'].mean())

print("Missing Values Handled Successfully")
```

```python
df.to_csv('fitness_data.csv', index=False)

print("Dataset saved successfully!")
```

```python
print("Steps Data:")
print(df['Steps'])

print("\nSteps and Calories:")
print(df[['Steps', 'Calories']])
```

```python
print("First 5 Rows:")
print(df.iloc[0:5])

print("\nFirst Row:")
print(df.iloc[0])

print("\nSelected Columns:")
print(df.loc[0:4, ['Date', 'Steps', 'Calories']])
```

```python
active_days = df[df['Steps'] > 7000]

print("Days with More Than 7000 Steps:")
print(active_days)
```

```python
sorted_data = df.sort_values(
    by='Steps',
    ascending=False
)

print("Fitness Data Sorted by Steps:")
print(sorted_data)
```

```python
df['Total_Activity'] = (
    df['Steps'] + df['Exercise_Minutes']
)

df['Calories_Per_Step'] = (
    df['Calories'] / df['Steps']
)

print(df.head())
```

```python
print("Fitness Activity Summary:")

print(df.groupby('Exercise_Minutes')['Calories'].mean())

print("\nStatistical Aggregation:")

print(df[['Steps', 'Calories', 'Sleep_Hours']].agg(
    ['mean', 'max', 'min']
))
```

```python
df['Activity_Level'] = df['Steps'].apply(
    lambda x: 'High' if x >= 8000 else 'Low'
)

pivot = pd.pivot_table(
    df,
    values='Calories',
    index='Activity_Level',
    aggfunc='mean'
)

print("Average Calories by Activity Level:")
print(pivot)
```

```python
df['Activity_Level'] = df['Activity_Level'].str.upper()

print(df[['Steps', 'Activity_Level']].head())
```

```python
import pandas as pd

data1 = df.head(5)

data2 = df.tail(5)

combined_data = pd.concat(
    [data1, data2],
    ignore_index=True
)

print("Combined Dataset:")
print(combined_data)
```

```python
health_data = pd.DataFrame({
    'Date': df['Date'],
    'Heart_Rate': np.random.randint(60, 100, 30)
})

merged_data = pd.merge(
    df,
    health_data,
    on='Date'
)

print("Merged Dataset:")
print(merged_data.head())
```

```python
fitness = df[['Steps', 'Calories']].head(5)

sleep = df[['Sleep_Hours', 'Exercise_Minutes']].head(5)

joined_data = fitness.join(sleep)

print("Joined Dataset:")
print(joined_data)
```

```python
print("Mean:")
print(df['Steps'].mean())

print("\nMedian:")
print(df['Steps'].median())

print("\nMode:")
print(df['Steps'].mode())

print("\nStandard Deviation:")
print(df['Steps'].std())

print("\nMinimum:")
print(df['Steps'].min())

print("\nMaximum:")
print(df['Steps'].max())
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df['Steps_Scaled'] = scaler.fit_transform(
    df[['Steps']]
)

print(df[['Steps', 'Steps_Scaled']].head())
```

```python
from sklearn.preprocessing import StandardScaler

standard = StandardScaler()

df['Steps_Standardized'] = standard.fit_transform(
    df[['Steps']]
)

print(df[['Steps', 'Steps_Standardized']].head())
```

```python
steps = np.sort(df['Steps'].values)

n = len(steps)

gini = (
    2 * np.sum((np.arange(1, n + 1)) * steps)
    / (n * np.sum(steps))
) - ((n + 1) / n)

print("Gini Coefficient:", gini)
```
